# **Problem Statement**

---



**Multi-line sequence fasta to single-line fasta file**

Fasta file format is shown below. Many times, it is annoying to find sequences of interest because of fix-width format, i.e., if 60 character fix-width file is there then only except the header (starting with >) of the sequence each line will have 60 characters. So, if a sequence is of width 600, then it would be written in 10 lines. Your goal is convert this format to a format with header in a single line followed by its sequence in a single line.

# **Approach**

---



FASTA files often store sequences with fixed-width lines, making searches across line breaks difficult. This code reads each line, checks for headers (lines starting with >), and writes them normally. For sequence lines, it concatenates them into a single line per header without line breaks. As a result, each sequence appears in a single continuous line, immediately after its header, improving readability and enabling direct pattern matching.

```python
# Convert multi-line FASTA to single-line per sequence format

input_path = "multiline_input.fasta"
output_path = "q3_output.fasta"

with open(input_path, 'r') as infile, open(output_path, 'w') as outfile:
    first_header = True  # Track whether the first header has been written

    for raw_line in infile:
        line = raw_line.strip()

        if line.startswith('>'):
            if not first_header:
                outfile.write('\n')  # Finish previous sequence
            outfile.write(line + '\n')  # Write new header
            first_header = False
        else:
            outfile.write(line)  # Concatenate sequence lines

    outfile.write('\n')  # Final newline to properly terminate last sequence

print(f"Reformatted file saved as: '{output_path}'")
```
